```python
# Import libraries
import plotly.express as px # to create interactive figure
import plotly  # to show the default renderer
```

```python
# See all renderer parameters
plotly.io.renderers
```

```python
# See the default renderer in current environment (colab)
plotly.io.renderers.default  # in colab, default renderer is colab
```

```python
# Creating a sample interactive figure
fig = px.bar(x=["a", "b", "c"], y=[1, 3, 2], height=300, width=500)
fig.show()  # in colab, default renderer is 'colab'

# this default renderer render figures well within colab interface,
# BUT after exporting to html, it doesn't work in the html file,
# because we're using jupyter nbconvert, it would call jupyter first,
# then render it and convert to html
```

```python
fig.show(renderer="notebook")

# this renderer is jupyter notebook renderer, which shows figures well
# in jupyter, but can't render in colab

# this renderer used to work on showing plots in exported html files,
# but not working now, don't know why
```

```python
fig.show(renderer="notebook_connected")

# this is also jupyter notebook renderer, and can't show figures in colab
# BUT, this renderer works after exporting ipynb to html files
```
